```python
# 1. Perform One-Hot Encoding as you were doing
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
if 'policy_id' in categorical_cols:
    categorical_cols.remove('policy_id')

df_one_hot = pd.get_dummies(df.drop(columns=['policy_id']), columns=categorical_cols)

# 2. CALCULATE CORRELATION TO REDUCE COLUMNS
# We find the correlation of every column with 'is_claim'
correlations = df_one_hot.corr()['is_claim'].abs().sort_values(ascending=False)

# 3. SELECT TOP 88 COLUMNS
# This takes the 88 columns with the highest relationship to 'is_claim'
# (Includes 'is_claim' itself, leaving 87 features)
top_88_features = correlations.head(88).index.tolist()
df_final = df_one_hot[top_88_features]

# 4. Verify the count
print(f"Original Columns with dummies: {df_one_hot.shape[1]}")
print(f"Reduced Columns: {df_final.shape[1]}")

# Save the final reduced dataset
df_final.to_csv('final_reduced_data.csv', index=False)
df_final.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=123456)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.fit_transform(x_test)
```

```python
#SMOTE 
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
x_train_sm, y_train_sm = smote.fit_resample(x_train, y_train)

```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(max_iter=1000, C=0.3)
lr.fit(x_train_scaled,y_train)
y_pred_lr = lr.predict(x_test_scaled)
```

```python
from sklearn.linear_model import LogisticRegression
def create_model_logistic_regression():

 lr = LogisticRegression(max_iter=1000, C=0.3,solver='liblinear')
 lr.fit(x_train_sm,y_train_sm)
 return lr
```

```python
from sklearn.neighbors import KNeighborsClassifier
def create_model_knn():

 knn = KNeighborsClassifier(n_neighbors=100)
 knn.fit(x_train_scaled,y_train)
 return knn
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier()
knn.fit(x_train_scaled,y_train)
y_pred_knn = knn.predict(x_test_scaled)
```

```python
from sklearn.svm import SVC 
def create_model_svm():

 model = SVC()
 model.fit(x_train_scaled,y_train)
 return model   
```

```python
from sklearn.svm import SVC
model = SVC()
model.fit(x_train_scaled,y_train)
y_pred_svm = model.predict(x_test_scaled)  

```

```python
from sklearn.tree import DecisionTreeClassifier
def create_model_decision_tree():

 dt = DecisionTreeClassifier(random_state=42)
 dt.fit(x_train_scaled, y_train) 
 return dt   
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(random_state=42)
dt.fit(x_train_scaled, y_train)
y_pred_dt = dt.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score


```

```python
#Evaluate model performance
def evaluate_model(y_test, y_pred):
    return {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, average='macro'),
        "Recall": recall_score(y_test, y_pred, average='macro'),
        "F1 Score": f1_score(y_test, y_pred, average='macro')
    }
```

```python
results = pd.DataFrame({
    "Logistic Regression": evaluate_model(y_test, y_pred_lr),
    "KNN": evaluate_model(y_test, y_pred_knn),
    "SVM": evaluate_model(y_test, y_pred_svm),
    "Decision Tree": evaluate_model(y_test, y_pred_dt),

    }).T

results
```

```python
models = [
    ("Logistic Regression", create_model_logistic_regression(), (0, 0)),
    ("KNN", create_model_knn(), (0, 1)),
    ("SVM", create_model_svm(), (1, 0)),
    ("Decision Tree", create_model_decision_tree(), (1, 1)),

]


```

```python
results_df = pd.DataFrame(results).sort_values(by="Accuracy", ascending=False)
print("\nBest Performing Model:\n")
print(results_df)
```

```python
fig,axs = plt.subplots(2,2,figsize=(15, 20))
for model_name, model, position in models:
    y_pred = model.predict(x_test)
    axs[position[0], position[1]].scatter(
       x_test['age_of_car'][y_pred==0],x_test['age_of_policyholder'][y_pred==0],
       color='green',label='Wont Suffer'
    )
    axs[position[0],position[1]].scatter(
        x_test['age_of_car'][y_pred==1],x_test['age_of_policyholder'][y_pred==1],
        color='red',label='Will Suffer'
    )

    axs[position[0],position[1]].set(xlabel='age_of_car', ylabel='age_of_policyholder')
    axs[position[0],position[1]].set_title(model_name)
```

```python
#Ratio
import re
import numpy as np

def extract_torque_rpm(value):
    try:
        torque = float(re.findall(r'[\d.]+', value)[0])
        rpm = float(re.findall(r'@([\d]+)', value)[0])
        return torque, rpm
    except:
        return np.nan, np.nan

df[['torque', 'torque_rpm']] = df['max_torque'].apply(
    lambda x: pd.Series(extract_torque_rpm(str(x)))
)
```

```python
def extract_power_rpm(value):
    try:
        power = float(re.findall(r'[\d.]+', value)[0])
        rpm = float(re.findall(r'@([\d]+)', value)[0])
        return power, rpm
    except:
        return np.nan, np.nan

df[['power', 'power_rpm']] = df['max_power'].apply(
    lambda x: pd.Series(extract_power_rpm(str(x)))
)
```

```python
df['torque_to_rpm_ratio'] = df['torque'] / df['torque_rpm']


df['power_to_rpm_ratio'] = df['power'] / df['power_rpm']
```
