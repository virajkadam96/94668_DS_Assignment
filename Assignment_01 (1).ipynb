### 1. Write a NumPy program to create an array with the values 1, 7, 13, 105    and determine the size of the memory occupied by the array.


```python
import numpy as np

arr = np.array([1, 7, 13, 105])
print("Array:", arr)
print("Memory size (bytes):", arr.nbytes)
```

### 2. Write a NumPy program to create an element-wise comparison (greater,    greater_equal, less and less_equal) of two given arrays.

```python
a = np.array([10, 20, 30])
b = np.array([15, 20, 25])

print("Greater:", a > b)
print("Greater Equal:", a >= b)
print("Less:", a < b)
print("Less Equal:", a <= b)
```

### 6. Write a NumPy program to get the element-wise remainder of an array of numbers from 20 to 50 which are divisible by 3.

```python
arr = np.arange(20, 51)
div_by_3 = arr[arr % 3 == 0]
remainder = div_by_3 % 3

print("Numbers:", div_by_3)
print("Remainders:", remainder)
```

### 7. Write a Pandas program to add, subtract, multiple and divide two Pandas Series.
   Sample Series: [2, 4, 6, 8, 10], [1, 3, 5, 7, 9]

```python
import pandas as pd

s1 = pd.Series([2, 4, 6, 8, 10])
s2 = pd.Series([1, 3, 5, 7, 9])

print("Addition:\n", s1 + s2)
print("Subtraction:\n", s1 - s2)
print("Multiplication:\n", s1 * s2)
print("Division:\n", s1 / s2)
```

### 11. Consider mtcars.csv file  and plot the following graphs.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("mtcars.csv")
print(data.head())
```

### i) Line plot

```python
plt.plot(data['mpg'])
plt.title("Line Plot of MPG")
plt.xlabel("Index")
plt.ylabel("MPG")
plt.show()
```

### ii) Bar Graph

```python
plt.bar(data.index, data['cyl'])
plt.title("Bar Graph of Cylinders")
plt.show()
```

### iii) Pie Chart


```python
data['cyl'].value_counts().plot.pie(autopct='%1.1f%%')
plt.title("Pie Chart of Cylinders")
plt.show()
```

### iv) Boxplot


```python
sns.boxplot(data['mpg'])
plt.title("Boxplot of MPG")
plt.show()
```

### v) Pairplot


```python
sns.pairplot(data)
plt.show()
```

###  vi) Scatter Plot


```python
plt.scatter(data['wt'], data['mpg'])
plt.xlabel("Weight")
plt.ylabel("MPG")
plt.title("Scatter Plot")
plt.show()
```

### vii) Histogram


```python
plt.hist(data['mpg'], bins=10)
plt.title("Histogram of MPG")
plt.xlabel("MPG")
plt.ylabel("Frequency")
plt.show()
```
