Create a Simple Linear Regression model for a given dataset. Also plot regression line.

Datasets:
1. NewspaperData.csv
Independent = ‘Daily’, Dependent=’Sunday’


```python
newspaper_data = pd.read_csv('C:\\Users\\lenovo\\Desktop\\DSML\\IIT-08-A-DS\\Datasets\\NewspaperData.csv')
print("Dataset shape:", newspaper_data.shape)
print("\nFirst few rows:")
print(newspaper_data.head())
```

```python
X1 = newspaper_data[['daily']].values
y1 = newspaper_data['sunday'].values
```

```python
model1 = LinearRegression()
model1.fit(X1, y1)
```

```python
y1_pred = model1.predict(X1)
```

```python
r2_1 = r2_score(y1, y1_pred)
rmse_1 = np.sqrt(mean_squared_error(y1, y1_pred))
```

```python
print(f"\nModel Equation: Sunday = {model1.intercept_:.4f} + {model1.coef_[0]:.4f} * Daily")
print(f"R² Score: {r2_1:.4f}")
print(f"RMSE: {rmse_1:.4f}")
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X1, y1, alpha=0.6, label='Actual Data')
plt.plot(X1, y1_pred, color='red', linewidth=2, label='Regression Line')
plt.xlabel('Daily Sales')
plt.ylabel('Sunday Sales')
plt.title('Newspaper Data: Daily vs Sunday Sales')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('newspaper_regression.png', dpi=300)
plt.show()
```

Consider below columns for prdecting the price.
("Age_08_04","KM","HP","cc","Doors","Gears","Quarterly_Tax","Weight")



```python
toyota_data = pd.read_csv('C:\\Users\\lenovo\\Desktop\\DSML\\IIT-08-A-DS\\Datasets\\ToyotaCorolla.csv', encoding='latin1')
print("Dataset shape:", toyota_data.shape)
print("\nFirst few rows:")
print(toyota_data.head())
```

```python
features = ["Age_08_04", "KM", "HP", "cc", "Doors", "Gears", "Quarterly_Tax", "Weight"]
X_toyota = toyota_data[features]
y_toyota = toyota_data['Price']
```

```python
print("\nFeatures used:", features)
print("\nFeature statistics:")
print(X_toyota.describe())
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_toyota, y_toyota, test_size=0.2, random_state=42)
```

```python
model_toyota = LinearRegression()
model_toyota.fit(X_train, y_train)
```

```python
y_train_pred = model_toyota.predict(X_train)
y_test_pred = model_toyota.predict(X_test)
```

```python
r2_train = r2_score(y_train, y_train_pred)
r2_test = r2_score(y_test, y_test_pred)
rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))
```

```python
print(f"\n{'='*60}")
print("MODEL PERFORMANCE")
print(f"{'='*60}")
print(f"Training R² Score: {r2_train:.4f}")
print(f"Testing R² Score: {r2_test:.4f}")
print(f"Training RMSE: {rmse_train:.2f}")
print(f"Testing RMSE: {rmse_test:.2f}")

print(f"\nIntercept: {model_toyota.intercept_:.2f}")
print("\nCoefficients:")
for feature, coef in zip(features, model_toyota.coef_):
    print(f"  {feature:20s}: {coef:10.4f}")
```

```python
plt.subplot(1, 2, 2)
plt.scatter(y_test, y_test_pred, alpha=0.5)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Test Set: Actual vs Predicted')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('toyota_price_prediction.png', dpi=300)
plt.show()
```
